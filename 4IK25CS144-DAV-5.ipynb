```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the dataset
df = pd.read_csv('StudentsPerformance.csv')

print("First 5 rows of the dataset:")
print(df.head())
print("\nDataset shape:")
print(df.shape)
```

```python
data=df["math score"].dropna()
print("\nNumber of math score values:",len(data))
```

```python
# Calculate Q1 (25th percentile)
Q1 = data.quantile(0.25)
# Calculate Q3 (75th percentile)
Q3 = data.quantile(0.75)
# Calculate Interquartile Range
IQR = Q3 - Q1
# Calculate lower limit
lower_limit = Q1 - 1.5 * IQR
# calculate upper limit
upper_limit = Q3 + 1.5 * IQR

print("\nIQR METHOD")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
# Find IQR outliers
iqr_outliers = df[
    (df["math score"] < lower_limit) | (df["math score"] > upper_limit)
]
print("Number of IQR outliers:", len(iqr_outliers))
```

```python
mean = data.mean()
std = data.std()
df["Z_score"] = (df["math score"] - mean) / std
z_outliers = df[abs(df["Z_score"]) > 3]
print("\nZ-SCORE METHOD")
print("Mean:", mean)
print("Standard Deviation:", std)
print("Number of Z-score outliers:", len(z_outliers))
```

```python
print("\nIQR OUTLIERS")
if len(iqr_outliers) > 0:
  print(iqr_outliers[["math score"]])
else:
  print("No IQR outliers found.")
print("\n Z-SCORE OUTLIERS")
if len(z_outliers) > 0:
  print(z_outliers[["math score", "Z_score"]])
else:
  print("No Z-score outliers found")
```

```python
df["math score_treated"] = df["math score"].clip(lower=lower_limit, upper=upper_limit)
print("Treatment method: Capping using IQR limits")
print("\nBEFORE TREATMENT")
print(df["math score"].describe())
print("\nAFTER TREATMENT")
print(df["math score_treated"].describe())
```

```python
plt.figure(figsize=(12, 5))

# Before treatment
plt.subplot(1, 2, 1)
plt.boxplot(df["math score"].dropna())
plt.title("Math Score - Before Treatment")
plt.ylabel("Math Score")

# After treatment
plt.subplot(1, 2, 2)
plt.boxplot(df["math score_treated"].dropna())
plt.title("Math Score - After Treatment")
plt.ylabel("Math Score")

plt.tight_layout()
plt.show()
```
